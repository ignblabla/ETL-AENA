# Patrón General: Los 5 pasos del Análisis Exploratorio Inicial

En cualquier proyecto de datos, antes de filtrar, graficar o modelar, siempre se ejecuta la misma secuencia diagnóstica para responder a **tres preguntas fundamentales**:

* 📊 **¿Cuántos datos tengo?**
* 🏷️ **¿Cómo interpreta pandas cada columna?**
* 🔍 **¿Qué aspecto real tienen los primeros registros?**

---

> 💡 **Objetivo:** Establecer un diagnóstico rápido y estructurado del dataset antes de aplicar transformaciones, limpieza o modelado.

```python
import pandas as pd

# 1. Carga de datos
df = pd.read_csv("../data/historico_vuelos.csv")

# 2. Diagnóstico estructural
print("Dimensiones (filas, columnas):", df.shape)

# 3. Tipos de datos
print("\nTipos por columna:\n", df.dtypes)

# 4. Radiografía completa y nulos
print("\nInformación general:")
df.info()

# 5. Vista previa
print("\nPrimeras 5 filas:")
df.head()
```

# Convertir un string fecha a datetime

Una vez cargado el DataFrame, realizamos la conversión.

```python
df["salida_programada"] = pd.to_datetime(df["salida_programada"], errors="coerce")
df["salida_estimada"] = pd.to_datetime(df["salida_estimada"], errors="coerce")
```

# Contar valores únicos por columna

```python
# Contar cuántos valores únicos hay en cada columna
print("Valores únicos por columna:")
print(df.nunique())
```

```python
# Contar cuántas veces aparece cada valor único
print("\nRegistros por aeropuerto:")
print(df["aeropuerto"].value_counts())
```
